```python
numbers = [23, 56, 12, 89, 45, 67]
loc = numbers[0]

for i in numbers:
        if i >loc:
            loc = i
print(loc)
    
```

8. Nested for loop

Print this pattern:

*
**
***

*****

You must use nested for loops.

```python
for i in range(1,6):
    for j in range(1,i+1):
        print("*",end = "")
    print()
```

```python
for i in range(6,1,-1):
    for j in range(1,i):
        print("*",end = " ")
    print()
```

9. for + if + counter

Given:

numbers = [5, 12, 8, 21, 30, 7, 14, 3]

Count how many numbers are even and how many are odd.

Expected:

Even: 4
Odd: 4

Don't use count().

```python
numbers = [5, 12, 8, 21, 30, 7, 14, 3]
even = 0
odd = 0

for i in numbers:
    if i%2==0:
        even+=1
    else :
        odd+=1
        
print("Even:",even)
print("Odd:",odd)
```

10. 🔥 Challenge — second largest

Given:

numbers = [10, 45, 23, 89, 67, 89, 34, 56]

Find the second largest unique number using a for loop.

Expected:

67

Rules:

Don't use sort()
Don't use max()
Don't use set()
Don't use sorted()

```python
numbers = [10, 45, 23, 89, 67, 89, 34, 56]

lar = numbers[0]
sec_lar = numbers[0]

for i in numbers:
    if i > lar:
        sec_lar = lar
        lar = i
    elif i > sec_lar and i != lar:
        sec_lar = i

print(sec_lar)
```
