2.Even or Odd + Positive or Negative

Take an integer.

First determine whether it is positive, negative, or zero.

If it is not zero, additionally determine whether it is even or odd.

Example:

Enter number: -7
Output:
Negative
Odd

```python
user1 = int(input("Enter number:"))

if user1 !=0 :    
    if user1 %2==0 and user1>0:
        print("Positive")
        print("Even")
    elif user1 %2==0 and user1<0:
        print("Negative")
        print("Even")
    elif user1 %2!=0 and user1>0:
        print("Positive")
        print("Odd")
    else :
        print("Negative")
        print("Odd")
else:
    print("Zero")
    
```

4. Student Grade Calculator

Take marks from 0–100.

Use:

90–100 → A
80–89  → B
70–79  → C
60–69  → D
50–59  → E
Below 50 → Fail

Also handle invalid marks:

marks < 0 or marks > 100
→ Invalid marks

This tests if + elif + else.

```python
marks  = int(input("Enter your marks:" ))

if marks <0 or marks >100:
    print("Invalid marks")
else:
    if marks >=90 and marks <=100:
        print("A")
    elif marks >=80 and marks <=89:
        print("B")
    elif marks >=70 and marks <=79:
        print("C")
    elif marks >=60 and marks <=69:
        print("D")
    elif marks >=50 and marks <=59:
        print("E")
    else:
        print("Fail")
        
```

6. Electricity Bill — Real-World Problem

Take the number of units consumed.

Calculate the bill:

0–100 units       → ₹5 per unit
101–200 units     → ₹7 per unit
201–300 units     → ₹10 per unit
Above 300 units   → ₹15 per unit

Example:

Units: 250

Calculate the appropriate bill.

⚠️ Think carefully about whether the rate applies to all units or only the units
within each slab. Choose a reasonable slab-based interpretation and implement it consistently.

```python
ebill = int(input("Enter the units consumed:"))

if ebill <0:
    print("Check and re-enter the correct value")
else:
    if ebill >=0 and ebill<=100:
        print("₹5 per unit ,so the amount to be paid is ₹",ebill*5)
    elif ebill >=101 and ebill<=200:
        print("₹7 per unit ,so the amount to be paid is ₹",ebill*7)
    elif ebill >=201 and ebill<=300:
        print("₹10 per unit ,so the amount to be paid is ₹",ebill*10)
    else:
        print("₹15 per unit ,so the amount to be paid is ₹",ebill*15)
```

7. Login System — Nested if

Create:

username = "guru"
password = "python123"

Ask the user for username and password.

Logic:

Correct username?
    ↓
    Yes → Check password
              ↓
              Correct → Login successful
              Wrong   → Wrong password

    No → Invalid username

This is your first nested if challenge.

```python
username = input("Enter your Username:")
password = input("Enter your Password:")

if username == "guru":
    if password == "python123":
        print("Login successful")
    else:
        print("Wrong password")
else:
    print("Invalid username")
```

```python
8. ATM Withdrawal — Nested if + and

Take:

balance
withdraw_amount

Rules:

Withdrawal amount must be greater than 0.
Withdrawal amount must be a multiple of 100.
Withdrawal amount must not exceed the balance.
If everything is valid → withdrawal successful.
Otherwise display the correct reason.

Example:

Balance: ₹10000
Withdraw: ₹2500

Output: Withdrawal successful
Remaining balance: ₹7500
```

```python
balance =  10000
withdraw_amount = int(input("Enter amount to withdraw:"))

if withdraw_amount >0 :
    if  withdraw_amount %100==0:        
        if withdraw_amount <= balance:
            balance-=withdraw_amount
            print(" Withdrawal successful")
            print("Remaining balance: ₹",balance)
        else:
            print("⚠️Amount is exceed from your balance")
    else :
        print("⚠️Amount must Multiple of 100")
        
else:
    print("⚠️Check the Amount correctly")
```

```python
9. Triangle Validator + Type

Take three sides:

a
b
c

First determine whether they can form a triangle:

a + b > c
a + c > b
b + c > a

If valid, determine whether it is:

Equilateral → all 3 equal
Isosceles   → any 2 equal
Scalene     → all different

Example:

3 3 3
→ Valid Triangle
→ Equilateral
```

```python
a = int(input("Enter A len:"))
b = int(input("Enter B len:"))
c = int(input("Enter C len:"))

if a+b>c and a + c > b and b + c > a:
    print("Valid Triangle")
    if a == b and b == c:
        print("Equilateral")
    elif a==b or b==c or c==a :
        print("Isosceles")
    else :
        print("Scalene")
else :
    print("Not an Triangle")
    
    
```
