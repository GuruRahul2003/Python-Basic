```python
# Operators 

# Arithmetic Operator 
a = 10
b = 2
c = 2

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a**b)
print(a//b)
print(a%b) # remainder
```

```python
# Logical Operator

# and ,or, not , can be use in decision making statement

a1 = 21 
a2 = 17
adult = True 

print(a1 >18 and adult) 

print(a2 >18 and adult) # both must be True 

print(a2 >18 or adult) # any one need to be True 

print(not adult ) # Give opposite answer
```

```python
# Comparison Opeerator  , gives Boolean answers 

# >,<,==,<=,>=,!=

print(a1 == a2)
print(a1 != a2)
print(a1 >= a2)
print(a1 <= a2)
```

```python
#  Assignment Operator , mainly used for increasing and decreasing the value 

# +=,-=,*=,/=,=

count = 1
count+=1 # increase the value one same for all sign 
print(count)
```

```python
# Membership operator 

# in , not in 

A = "Hi , How are you"

print("Hi" in A)
print("HI" in A)
print("HI" not in A)
```

```python
# Identity Operator 

# is , is not 

a = 1234
b= 1234

print(id(a))
print(id(b))

print(a is b) # it checks the memory location 
```

```python
# Bitwise operator 
g = 12
r = 4
print(g<<r) # it will work like 12 * 2^4 
print(g>>r) # 12 / 2^4
print(12*(2**4))
print(12/(2**4))
```
