```python
# String Functions 

gr  = "  Hi I'm Gr   "
print(gr.casefold())  # converts string to lowercase  same operations as lower but useful for comparison 
print(gr.lower()) # converts string to lowercase 
print(gr.upper()) # converts to upper
print(gr.capitalize()) # convert the first letter to capital 
print(gr.title()) # convert every 1st char to capital 
print(gr.find("G")) # find the index of the character 
print(gr.isupper()) # check all char is upper 
print(gr.islower()) # check all char is lower 
print(gr.strip()) # remove unwanted space from beginning and in end 
print(gr.rstrip()) # remove unwanted space from right side
print(gr.lstrip()) # remove unwnated space in left side 
print(gr.startswith(" ")) # check the starting char
print(gr.endswith("r")) #  check the ending char
print(len(gr)) # gives us the length 
print(gr.replace("Gr","Guru" )) # replacing the char 
print(gr.count("r")) # count the char 
print(gr.center(50)) # give equal space at start and end 
print(gr.ljust(50)) # leave space in right side 
print(gr.rjust(50)) # give space in left side 
print(gr.isalnum()) # it must only sequence of char with no space 
print(gr.rfind("G")) # it gives the end char 
print(gr.swapcase()) # low to up , up to low case 
print(gr.partition("m")) # will break into exactly 3 parts 
gr2 = "Hi-Guru-How-are-you"
print(gr2.split("-")) # convert the string to list by separator 
gr1 = ["Hi","Bye","Hello"]
print("--".join(gr1)) # converts list to string 
print(gr.index("m"))
```

```python
# Idexing and Slicing 

guru = "Hi everyone use the resources to excel in python foundation"
# indexing
print(guru[5])
print(guru[7])

# slicing 
print(guru[0:11])
# negative slicing or Travsing
print(guru[::-1])
print(guru[11:0:-1])
```

```python
# String Formating  

guru = "Hi everyone use the resources to excel in python foundation"
guru1 = "and try to others whom you feel will helpful  "

# formated string 
print(f"{guru} {guru1} ")

# general string
print(guru,"",guru1)

```
