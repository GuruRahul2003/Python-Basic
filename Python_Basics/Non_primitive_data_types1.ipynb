# Python Non primitive Data types 

## List and Tuple 


```python
# list is Mutable and allow Duplicates 

# Syntax 

lis = ["Gr", "Hari", "Jayan","Sathish"]
lis
# list methods -> append, extend , insert , pop, remove ,clear ,copy, reverse, sort, index , count 

# ----------------------Adding values to list-------------------------------------#

lis.append("Guru") # add single value at last 
print(lis) 

lis1 = ["A","B","C"]
lis.extend(lis1) # add multiple value at last 
print(lis) 

lis.insert(0,"gr7") # insert value at specfic location 
print(lis) 

# ----------------------Removing values from list-------------------------------------#

new_id = lis.pop(0) # remove that value and store in the separate variable  , bydefault remove the last value or the mentioned index value 
print(new_id)

lis.remove("A") # remove the specified value from the list 
print(lis)

#lis.clear()  # clear the entire list 
print(lis)

```

```python
# ---------------------- list most used methods -------------------------------------#

print(lis.count("B")) # count the given value 
print(len(lis)) # will shows the length of the list
print(lis.index("Guru")) # will show the index of the given value 
lis.sort()
print(lis)

lis.sort(reverse = True)
print(lis)

lis = ["Gr", "Hari","Sathish", "Jayan"]
lis.reverse()
print(lis)

sor_lis = sorted(lis)
sor_lis
```

```python
# ----------------------creating duplicate list-------------------------------------#

# shallow copy 

lis3 = lis # this is shallow copy what ever changes made in the duplicate affect the main list 

# lis3.remove("Jayan")
print(lis3)
print(lis) # changes occurs in both 

# deep copy 

lis4 = lis1.copy()
lis4.remove("A")
print(lis4)
print(lis1)  # changes that happen in duplicates will not affect the orginal 
```

```python
# ----------------------fetching values from list-------------------------------------#

# index  

new_lis = ["Gr", "Hari", "Jayan","Sathish"]
print(new_lis[0])
print(new_lis[2])

# slicing

print(new_lis[0::2])
print(new_lis[::-1])
print(new_lis[4:0:-1])
```

```python
# Tuple is immutable and allow duplicates , supports indexing and slicing , only have count and index 

tup = ("Hello,","use" ,"this","to","develop") 
tup1 = ("your knowledge",)
print(tup+tup1)
print(tup[2])
print(tup[0::2])
```
